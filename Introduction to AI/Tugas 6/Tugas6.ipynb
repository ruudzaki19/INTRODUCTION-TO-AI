```python
import pandas as pd

data = pd.DataFrame({
    'Harga': ['Murah', 'Murah', 'Sedang', 'Mahal', 'Mahal', 'Sedang', 'Sedang', 'Mahal'],
    'Diskon': ['Ada', 'Ada', 'Ada', 'Tidak Ada', 'Tidak Ada', 'Ada', 'Tidak Ada', 'Ada'],
    'Ulasan': ['Baik', 'Baik', 'Baik', 'Buruk', 'Buruk', 'Baik', 'Baik', 'Buruk'],
    'Hasil': ['Tertarik', 'Tertarik', 'Tertarik', 'Tidak Tertarik', 'Tidak Tertarik', 'Tertarik', 'Tertarik', 'Tidak Tertarik']
})

print(data)
```

```python
total_data = len(data)
prior_tertarik = len(data[data['Hasil'] == 'Tertarik']) / total_data
prior_tidak_tertarik = len(data[data['Hasil'] == 'Tidak Tertarik']) / total_data

print(f"Prior Tertarik: {prior_tertarik}")
print(f"Prior Tidak Tertarik: {prior_tidak_tertarik}")
```

```python
# Data baru yang diprediksi
input_harga = 'Sedang'
input_diskon = 'Ada'
input_ulasan = 'Baik'

# Likelihood untuk kelompok 'Tertarik'
df_t = data[data['Hasil'] == 'Tertarik']
l_harga_t = len(df_t[df_t['Harga'] == input_harga]) / len(df_t)
l_diskon_t = len(df_t[df_t['Diskon'] == input_diskon]) / len(df_t)
l_ulasan_t = len(df_t[df_t['Ulasan'] == input_ulasan]) / len(df_t)

# Likelihood untuk kelompok 'Tidak Tertarik'
df_tt = data[data['Hasil'] == 'Tidak Tertarik']
l_harga_tt = len(df_tt[df_tt['Harga'] == input_harga]) / len(df_tt)
l_diskon_tt = len(df_tt[df_tt['Diskon'] == input_diskon]) / len(df_tt)
l_ulasan_tt = len(df_tt[df_tt['Ulasan'] == input_ulasan]) / len(df_tt)

# Perhitungan Posterior
post_tertarik = prior_tertarik * l_harga_t * l_diskon_t * l_ulasan_t
post_tidak_tertarik = prior_tidak_tertarik * l_harga_tt * l_diskon_tt * l_ulasan_tt

print(f"Skor Posterior Tertarik: {post_tertarik}")
print(f"Skor Posterior Tidak Tertarik: {post_tidak_tertarik}")
```

```python
if post_tertarik > post_tidak_tertarik:
    hasil = "TERTARIK"
else:
    hasil = "TIDAK TERTARIK"

print(f"Hasil Klasifikasi: {hasil}")
print("Alasan: Skor posterior 'Tertarik' lebih besar dibandingkan 'Tidak Tertarik'.")
```
