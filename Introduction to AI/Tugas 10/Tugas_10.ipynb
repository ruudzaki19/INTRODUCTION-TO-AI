```python
import numpy as np
from sklearn.linear_model import Perceptron

# 1. Input data dari soal (X = [IPK, SKS], y = Status)
X = np.array([
    [3.5, 22],
    [2.8, 18],
    [3.2, 20]
])
y = np.array([1, 0, 1])

# 2. Setup model Perceptron
model = Perceptron(max_iter=1, random_state=42)
model.fit(X, y)

# 3. Plot bobot (w1, w2) dan bias (b) sesuai instruksi soal
model.coef_ = np.array([[1.0, 1.0]])
model.intercept_ = np.array([-20.0])

# 4. Uji data pertama (IPK: 3.5, SKS: 22)
data_pertama = np.array([[3.5, 22]])

# Hitung nilai z di Python
z = (model.coef_[0][0] * data_pertama[0][0]) + (model.coef_[0][1] * data_pertama[0][1]) + model.intercept_[0]
print(f"Nilai z hasil kodingan: {z}")

# Cek hasil prediksi y
output = model.predict(data_pertama)
print(f"Output Prediksi (y): {output[0]}")
```
